```python
from modified_gauss_newton import modified_gauss_newton
from problem import ThurberProblem
import numpy as np

thurber = ThurberProblem()
starts = thurber.get_starting_points()

for name, x0 in starts.items():
    print(f"--- {name} ---")
    
    final_params = modified_gauss_newton(
        problem=thurber, 
        x0=x0, 
        M0=1.0, 
        M_search=2, 
        tol=1e-12,
        max_iter=100
    )
    
    residuals = thurber.F(final_params)
    rss = np.sum(residuals**2)
    
    print(f"Params: {final_params}")
    print(f"Final RSS: {rss:.10e}")
```
